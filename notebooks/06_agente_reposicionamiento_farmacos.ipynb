```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
from IPython.display import display, Markdown

# 1. Configuración el LLM
llm = ChatGoogleGenerativeAI(model="gemini-3.8-flash", temperature=0)

# 2. Los blancos terapéuticos
blancos_terapeuticos = """
1. Complejo APC/C
2. CDC20
3. PLK1
4. Aurora B
5. Hsp90 (estabilizador de proteína)
6. Vía de señalización p53
"""

# 3. Prompt del Agente
prompt_template = PromptTemplate.from_template("""
Eres un farmacólogo clínico experto en reposicionamiento de fármacos y oncología pediátrica.
El equipo investiga el Síndrome de Aneuploidía Variegada en Mosaico (MVA), causado por una deficiencia en la proteína BUB1B (punto de control mitótico SAC debilitado).

Tu tarea es buscar opciones de reposicionamiento para TODOS los siguientes blancos terapéuticos a la vez:
{targets}

RESTRICCIÓN CRÍTICA:
Solo debes proponer fármacos que ya estén APROBADOS por la FDA o EMA. No incluyas moléculas en fase preclínica.

Genera la respuesta estrictamente con UNA SOLA tabla Markdown que consolide los 6 blancos, utilizando estas columnas exactas:
| Blanco Terapéutico | Fármaco Aprobado | Mecanismo de Acción | Estado FDA/EMA | Perfil Pediátrico |
|---|---|---|---|---|

(Si no hay fármacos aprobados directos para un blanco, sugiere el aprobado que actúe más cerca en la vía compensatoria).
""")

# 4. Pipeline con StrOutputParser
parser = StrOutputParser()
cadena = prompt_template | llm | parser

print("⏳ Ejecutando pipeline agentico ...\n")

# 5. Ejecución del pipeline
respuesta_texto = cadena.invoke({"targets": blancos_terapeuticos})

resultados_markdown = "# Resultados del Agente de Reposicionamiento\n\n" + respuesta_texto

print("✅ Búsqueda finalizada con éxito.")
```

```python
with open("candidatos_reposicionamiento6blancos.md", "w", encoding="utf-8") as f:
    f.write(resultados_markdown)

print("Archivo 'candidatos_reposicionamiento.md' guardado en el panel izquierdo.")
print("Vista previa de la investigación generada:")
print("-" * 50)

display(Markdown(resultados_markdown))
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
from IPython.display import display, Markdown

# 1. Configuración del LLM
llm = ChatGoogleGenerativeAI(model="gemini-3.5-flash-lite", temperature=0)

# 2. Blancos terapéuticos propuestos traducidos al INGLÉS para optimizar la búsqueda
blancos_terapeuticos_en = [
    "APC/C complex",
    "CDC20",
    "PLK1",
    "Aurora B kinase",
    "Hsp90 (protein stabilizer)",
    "p53 signaling pathway"
]

# 3. Prompt bilingüe: Razonamiento en inglés, salida en español
prompt_template = PromptTemplate.from_template("""
You are a clinical pharmacologist expert in drug repurposing and pediatric oncology.
The team is investigating Mosaic Variegated Aneuploidy (MVA) syndrome, caused by a BUB1B deficiency (weakened SAC mitotic checkpoint).

Your task is to find repurposing options for the following therapeutic target: {target}

CRITICAL RESTRICTION:
You must ONLY propose drugs that are currently APPROVED by the FDA or EMA. Do not include preclinical molecules.
Perform your internal search and pharmacological reasoning using English medical literature, but strictly output your final response in SPANISH.

Genera la respuesta estrictamente con esta tabla Markdown en español:
| Fármaco Aprobado | Mecanismo sobre {target} | Estado FDA/EMA | Perfil de Seguridad Pediátrica |
|---|---|---|---|

(Si no hay fármacos aprobados directos para este blanco, sugiere el aprobado que actúe más cerca en la vía compensatoria).
""")

parser = StrOutputParser()
cadena = prompt_template | llm | parser

print("⏳ Ejecutando pipeline agentico (Búsqueda médica en Inglés -> Reporte en Español)...\n")

resultados_markdown = "# Resultados del Agente de Reposicionamiento\n\n"

for target in blancos_terapeuticos_en:
    print(f"Analizando literatura para: {target}...")
    respuesta_texto = cadena.invoke({"target": target})

    resultados_markdown += f"### Blanco Terapéutico: {target}\n"
    resultados_markdown += respuesta_texto + "\n\n"

print("\n✅ Búsqueda finalizada.")


with open("candidatos_reposicionamiento6blancosEn.md", "w", encoding="utf-8") as f:
    f.write(resultados_markdown)

print("💾 Archivo 'candidatos_reposicionamiento6blancosEn.md' actualizado con datos de alta precisión.")
print("-" * 50)
display(Markdown(resultados_markdown))
```
