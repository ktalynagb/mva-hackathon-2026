## A continuación arquitectura del Agente RAG Bilingüe. Realiza una extracción de datos médicos (DGIdb y ClinicalTrials)

Este pipeline automatiza la búsqueda de reposicionamiento de fármacos utilizando una estrategia de **Retrieval-Augmented Generation (RAG)** dividida en tres fases:

1. **Extracción (Retrieval) en Inglés:** Las funciones `buscar_dgidb` y `buscar_clinical_trials` consultan APIs internacionales en su idioma nativo. Esto es importante porque buscar términos médicos en español en estas bases de datos reduce drásticamente los aciertos.
2. **Inyección de Contexto:** Los datos crudos (interacciones gen-fármaco y ensayos pediátricos) se inyectan en un prompt estructurado diseñado para un "farmacólogo clínico experto".
3. **Generación Aumentada (LLM):** Se utiliza un modelo de Gemini para evaluar los datos bajo una restricción estricta (solo fármacos aprobados por FDA/EMA). El modelo procesa la evidencia médica en inglés, razona las vías compensatorias y traduce el reporte final para generar una tabla en **español**.

```python
# @title
import requests
import time
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
from IPython.display import display, Markdown

# 1. Configuración del LLM
llm = ChatGoogleGenerativeAI(model="gemini-3.5-flash-lite", temperature=0)

# 2. Mapeo de Blancos Terapéuticos (Mostrados en español) a Genes (Universales)
blancos_genes = {
    "Complejo APC/C": "ANAPC1",
    "CDC20": "CDC20",
    "PLK1": "PLK1",
    "Aurora B": "AURKB",
    "Hsp90": "HSP90AA1",
    "Vía de señalización p53": "TP53"
}

# 3. Herramientas de Extracción (Buscando en bases de datos en INGLÉS)
def buscar_dgidb(gen):
    """Extrae interacciones de DGIdb. Los nombres de fármacos devueltos están en inglés."""
    try:
        url = f"https://dgidb.org/api/v2/interactions.json?genes={gen}"
        res = requests.get(url, timeout=10).json()
        farmacos = []
        if "matchedTerms" in res and len(res["matchedTerms"]) > 0:
            interacciones = res["matchedTerms"][0].get("interactions", [])
            for inter in interacciones[:5]:
                farmacos.append(inter.get("drugName", "").capitalize())
        # Devuelve el contexto en inglés para la IA
        return ", ".join(filter(None, farmacos)) if farmacos else "No direct interactions found in DGIdb."
    except Exception as e:
        return f"API Error: {e}"

def buscar_clinical_trials(farmaco):
    """Consulta ClinicalTrials.gov usando términos en inglés (pediatric OR child)."""
    # Verifica los mensajes de error en inglés de la función anterior
    if not farmaco or "Error" in farmaco or "No direct interactions" in farmaco:
        return "N/A"
    try:
        primer_farmaco = farmaco.split(",")[0].strip()
        # Búsqueda estructurada en inglés para obtener mejores resultados
        url = f"https://clinicaltrials.gov/api/v2/studies?query.term={primer_farmaco}+AND+(pediatric+OR+child)&pageSize=1"
        res = requests.get(url, timeout=10).json()
        total = res.get("totalCount", 0)
        # Devuelve el contexto en inglés para la IA
        return f"{total} pediatric trials registered for {primer_farmaco}." if total > 0 else f"0 pediatric trials registered for {primer_farmaco}."
    except Exception as e:
        return "Trial data temporarily unavailable."

# 4. Prompt RAG (Instrucción estricta: Analizar datos en inglés, responder en ESPAÑOL)
prompt_template = PromptTemplate.from_template("""
Eres un farmacólogo clínico experto. El equipo investiga el Síndrome de Aneuploidía Variegada en Mosaico (MVA) por deficiencia en BUB1B.

Blanco Terapéutico: {target} (Gen: {gene})

CONTEXTO DE BASES DE DATOS EN TIEMPO REAL (Extraído en Inglés):
Candidatos en DGIdb: {dgidb_data}
Evidencia Pediátrica (ClinicalTrials.gov): {ct_data}

RESTRICCIÓN CRÍTICA:
Analiza el contexto en inglés proporcionado arriba. Solo propón fármacos APROBADOS por FDA/EMA. Si los fármacos del contexto no están aprobados, o si no hay datos directos, razona y sugiere el fármaco aprobado que actúe más cerca en la vía compensatoria mitótica.

Genera la respuesta estrictamente con esta tabla Markdown en ESPAÑOL:
| Fármaco Aprobado | Mecanismo sobre {target} | Estado FDA/EMA | Perfil de Seguridad Pediátrica |
|---|---|---|---|
""")

parser = StrOutputParser()
cadena = prompt_template | llm | parser

print("⏳ Ejecutando Pipeline RAG (Consultas en Inglés -> Análisis IA -> Reporte en Español)...\n")

resultados_markdown = "# Informe Farmacológico de Reposicionamiento (Pipeline RAG)\n\n"

# 5. Ejecución del orquestador
for target, gene in blancos_genes.items():
    print(f" Consultando APIs médicas para: {target} (Gen: {gene})...")

    # Retrieval de datos empíricos
    dgidb_data = buscar_dgidb(gene)
    ct_data = buscar_clinical_trials(dgidb_data)

    # Inyección de contexto y generación
    respuesta_texto = cadena.invoke({
        "target": target,
        "gene": gene,
        "dgidb_data": dgidb_data,
        "ct_data": ct_data
    })

    resultados_markdown += f"### Blanco Terapéutico: {target}\n"
    resultados_markdown += respuesta_texto + "\n\n"
    time.sleep(2)

print("\n Búsqueda RAG finalizada.\n")

# 6. Exportación
with open("candidatos_reposicionamiento_rag_En.md", "w", encoding="utf-8") as f:
    f.write(resultados_markdown)

print("Archivo 'candidatos_reposicionamiento_rag_En.md' guardado en el panel izquierdo.")
print("-" * 50)
display(Markdown(resultados_markdown))
```
